Visualization 1: %%% EDIT THIS MARKDOWN CELL%%%

```r
scatterplot_reg_wins_and_total_wins <- ggplot(final_tidy_diverse_data, aes(x = reg_wins, y = total_wins_pct, color = school)) +
  geom_point() +
  xlab("Regional Wins") +
  ylab("Total Percent of Wins") + scale_color_brewer(palette = "Set2") +
  theme(text = element_text(size = 12)) + ggtitle("Total Percentage of Wins by Regional Wins by School")
scatterplot_reg_wins_and_total_wins
```

I made this visualization to see what kind of relationship regional wins and total percentage of wins had and if they were positive. This can help me see if a trend even exists before thinking about prediction. It also lets me see if that is the case for all schools in my data set yet it is still hard to parse out exactly how this trend looks like for each schools; though we see a strong positive relationship generally between regional wins and total percentage of wins for the schools. We also see that the relationship is linear. 

```r
reg_wins_by_total_pct <- ggplot(final_tidy_diverse_data) +
    geom_line(aes(x = reg_wins, y = total_wins_pct, colour = school)) +
   labs(colour = "School Names") +
    xlab("Regional Wins") + 
    ylab("Total Percentage of Wins") +
    facet_grid(cols = vars(school))
options(repr.plot.width = 12, repr.plot.height = 10) 
reg_wins_by_total_pct
```

Since I saw that the relationship between regional wins and total percentage of wins was strong and linear, I now want to see if this linearity is similar for each school and more closely look at this trend for each school. As shown above, the relationship is generally similar per school where as regional wins increase, total percentage of wins also increases for that school. However, we can see some differences in the slopes with Auburn's being steeper than the rest.

I plan to use linear regression because my data is likely to be linear based on what I see with my exploratory analysis so far, so I will likely not run into too many limitations to do with curvilinear or non-linear (in general) data. Additionally, it will help me with interpretability as I can get a value for the slope and intercept unlike knn regression. Limitations include that I might underfit my data a little bit because for schools like "Auburn" and "Drake", the relationship between regional wins and the total percentage of wins are slightly curved at the top. Thus, it is not perfectly linear but my linear regression will not capture that. I will split the data into training and testing once (as I am not using k nearest neighbours to choose k) doing an initial split with strata set to total_wins_pct and splitting it by .7 for training, .3 for testing. Thus, I will only start my split after renaming my schools into numbers. There will be no cross-validation as I am using linear regressions, so we do not need to pick k. I will assess the model through finding the smallest RMSPE(ie. the squared difference between the predicted and true response value for each observation in my testing set).It is useful because if it is a very small value, that would indicate that my model does a good job predicting the total percentage of wins as my observations do not diverge too much (in terms of vertical distance) from where the points would truly be.
